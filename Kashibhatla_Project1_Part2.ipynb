## Creating the cleaned dataset via python 

I will be keeping or importing the 11 motivation based columns into my new dataframe, but I will also add the gender, race/ethnicity, age as well as the interest rate features, which pertain to my original five questions 

```python
# My dataset questions before EDA and cleaning, evaluating the dataset for the following questions 
# and removing any unnecessary columns that are not needed for the analysis.

# 1. Based on the non-majors survey dataset, how did motivation levels 
# influence students to take a computing course?
 
# 2. What was the gender distribution of students as non-majors in the computing course?
df_non_majors_2024["Gender"].value_counts()

# 3. What is the race split between the non-computing students, ethnic groups in terms of taking 
# a computing course? 
df_non_majors_2024["Race/ethnicity"].value_counts()

# 4. What is the age split between the non-computing students, in terms of taking a computing class? 
df_non_majors_2024['Age'].value_counts()

# 5. Was there a scaling rate, from 1-5 between different criteria such as gender, race and age in terms
# of interest rate, between the non-computing students taking a computing class? 
df_non_majors_2024.groupby(['On a scale of 1 to 5, with 1 being not at all interested and 5 being extremely interested, how interested are you '
                                                            'taking more courses in Computer Science, Information Technology or Game Development?'])[['Gender', 'Race/ethnicity', 'Age']].value_counts()

# Create a new dataframe with only the necessary columns for analysis
df_non_majors_cleaned = df_non_majors_2024[df_non_majors_2024.columns[2:13]].copy()

# Add the other columns necessary for analysis(Gender, Race/Ethnicity, Age, and the interest scale)
df_non_majors_cleaned['gender'] = df_non_majors_2024["Gender"]
df_non_majors_cleaned['ethnicity'] = df_non_majors_2024["Race/ethnicity"]
df_non_majors_cleaned['age'] = df_non_majors_2024["Age"]
df_non_majors_cleaned['interest rate'] = df_non_majors_2024['On a scale of 1 to 5, with 1 being not at all interested and 5 being extremely interested, how interested are you '
                                                            'taking more courses in Computer Science, Information Technology or Game Development?']
```
