## Import and Explore our cleaned non-survey dataset

    We will be importing the necessary python data visualization and data analysis libraries as well as reading in the cleaned non-survery dataset and exploring the cleaned non-survey dataset with the head() function

```python
# Import the libraries for basic data visualization and analysis
import pandas as pd
#%pip install pandas
import matplotlib.pyplot as plt

# Read in the cleaned non-majors survey data via path 
df_cleaned_survey = pd.read_csv("Non-Majors Survey Results - Fall 2024 Cleaned Version.csv")

# Display the first five rows via head() 
df_cleaned_survey.head()
```

    Exploring the dataset via descriptive statistics, column names, shape of the dataset, dtypes, head, tail, sample as well as info and missing values

```python
# Explore the dataset using pandas methods

# Explore descriptive statistics
print(df_cleaned_survey.describe())

# Explore the column names
print(df_cleaned_survey.columns)

# Explore the shape of the dataset
print(df_cleaned_survey.shape)

# Explore the dtypes of each column
print(df_cleaned_survey.dtypes)

# Explore the head of the dataset
print(df_cleaned_survey.head())

# Explore the tail of the dataset
print(df_cleaned_survey.tail())

# Explore the sample of the dataset
print(df_cleaned_survey.sample(5))

# Explore the info of the dataset
print(df_cleaned_survey.info())

# Explore the missing values of the dataset
print(df_cleaned_survey.isnull().sum())
```

## To perform basic data EDA on Question 1

        1. Based on the non-majors survey dataset, how did motivation levels influence students to take a computing course?

```python
# Create a new dictionary to isolate the Yes motivation tags only 
motivation_yes = (df_cleaned_survey.iloc[:, :11] == "Yes").sum()

# Rename the motivation_yes index to shorter labels for easier data visualization
motivation_yes.index = [ 
    "Degree Requirements", "Current Skills", "Career Advancement", 
    "Career Changes", "Professional Developments", "Job Displacements",
    "Relocation", "IT Certs", "Financials", "Personal Enrich", "Curiosity"
]
# Graph by bar plot
motivation_yes.plot(kind="bar", title = "Motivation(Student)")
plt.xlabel("Number of students")
plt.show()
```

## Question 1 Findings

    Based on these visualization findings, the motivation a student takes a computing class was the highest 
    when it was part of their degree requirements, with roughly 82 students. The least motivation(s) a student 
    has when it comes to taking to a computing class were the IT certifications, relocation and career changes motives. 

## To perform basic data EDA on Question 2

    2. What was the gender distribution of students as non-majors in the computing course?


```python
# Create a gender_bar variable that counts the gender columns only
gender_bar = df_cleaned_survey["gender"].value_counts()

# Create a bar graph to visualize the gender distribution via bar graphs
gender_bar.plot(kind="bar", title = "Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

## Question 2 Findings

    The gender distribution of students in computing course as non-majors is represented by a majority of 
    women(approximately 59 students) versus the men, standing at approximately 28 students. 

## To perform basic data EDA on Question 3

    3. What is the race split between the non-computing students, ethnic groups in terms of taking a computing course? 

```python
# Create a race_bar variable to isolate the columns by race, use the string split, 
# explode and and strip methods again to splice our data against redundancies
race_bar = df_cleaned_survey["ethnicity"].str.split(";").explode().str.strip().value_counts()

# Create a race bar bar graph to see the racial and ethnic distributions for 
# non-computing students taking a computing course
race_bar.plot(kind="bar", title = "Race and Ethnicity")
plt.xlabel("Race and Ethnicity")
plt.ylabel("Number of Students")
plt.show()
```

## Question 3 Findings

    Based on these visualization, we can conclude objectively that the white/caucasian race presented the majority 
    of the students taking a computing course as non-majors, versus the minority groups(Black, Asian, Multi-Racial and American Indian)
    comprising smaller representations. 

## To perform basic data EDA on Question 4

    4. What is the age split between the non-computing students, in terms of taking a computing class? 


```python
# Create an age bar variable to isolate only the age columns from the cleaned survey
age_bar = df_cleaned_survey["age"].value_counts()

# Create an age bar graph to visualize the age distributions with students 
# taking a computing class as non-CS majors
age_bar.plot(kind="bar", title = "Age")
plt.xlabel("Student Age")
plt.ylabel("Number of Students")
plt.show()
```

## Question 4 Findings

    Based on these visualization findings, we can conclude objectively that between ages 19-20, there was a 
    higher representation of students as non-majors taking a computing class versus all other age groups, 
    which shows an objectively lower representation. 

## To perform basic data EDA on Question 5

     5. Was there a scaling rate, from 1-5 between different criteria gender and age only in terms of interest rate, between the non-computing students taking a computing class? 


```python

# Use the groupby function to be able to assess the mean distribution of gender vs. the average interest
# rate from our dataset
df_cleaned_survey.groupby("gender")["interest rate"].mean().plot(kind="bar")

# Plot Gender against the average interest rate
plt.title("Average Interest by Gender only")
plt.ylabel("Rating from 1 to 5")
plt.show()


# Use the groupby function to be able to assess the mean distribution of age vs. the average interest
# rate from our dataset 
df_cleaned_survey.groupby("age")["interest rate"].mean().plot(kind="bar")

# Plot age against the average interest rate
plt.title("Average Interest by Age only")
plt.ylabel("Rating from 1 to 5")
plt.show()

```

## Question 5 Findings

    distributed evenly for non-majors taking a computing class. In fact, a similar distribution exists 
    with gender against the average interest rate, with a steeper dip with the distribution albeit.
